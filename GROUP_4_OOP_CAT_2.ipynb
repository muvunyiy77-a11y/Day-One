OOP concepts demonstrated: classes/objects, _init_, self, encapsulation

method overriding, polymorphism, composition.


# The base Musician class

This holds what's shared: *name* and *years of experience*. Experience is private (***_years_experience***) and exposed through a validated @property.

```
#1. Musician Class
class Musician:
  def __init__(self, name, years_experience):
    self.name = name
    self.years_experience = years_experience

  @property
  def name(self):
    return self._name

  @name.setter
  def name(self, name):
    self._name = name

  @property
  def years_experience(self):
    return self._years_experience

  @years_experience.setter
  def years_experience(self, years_experience):
    self._years_experience = years_experience

  def play(self):
    return "plays an instrument"
```

# ** Subclasses — Guitarist and Drummer**

Each adds one extra attribute and overrides play(). Both call super().__init__() so the parent's constructor (and its validation) still runs.

```
# Sub classes
class Guitarist(Musician):
  def __init__(self, name, years_experience, guitar_type):
    super().__init__(name, years_experience)
    self.guitar_type = guitar_type

  def play(self):
    return f"{self.name} shreds on a {self.guitar_type}"


class Drummer(Musician):
  def __init__(self, name, years_experience, kit_size):
    super().__init__(name, years_experience)
    self.kit_size = kit_size

  def play(self):
    return f"{self.name} pounds a {self.kit_size}-piece kit"
```

# The Band class (composition, not inheritance)

A Band has musicians — it doesn't become one. That's composition. It never checks whether a member is a Guitarist or Drummer; it just calls .play() and trusts each object to know how to respond (polymorphism).

```
#2.	Band class A Band holds a list of musicians
class Band:
    """A Band HAS musicians (composition) — it is not a Musician itself."""

    def __init__(self):
        self.members = []

    def add_member(self, musician):
        self.members.append(musician)

    def total_experience(self):
        return sum(member.years_experience for member in self.members)

    def perform(self):
        # Polymorphism: no type-checking needed — each member knows how to play().
        for member in self.members:
            print(member.play())
```

# **Demo script**

```
#Demo Script
if __name__ == "__main__":
    print("=" * 40)
    print("       MUSIC BAND MANAGER DEMO")
    print("=" * 40)

    band = Band()

    jimi = Guitarist("Jimi", 12, "Stratocaster")
    ringo = Drummer("Ringo", 15, "5")

    band.add_member(jimi)
    band.add_member(ringo)

    print("\n-- Performance --")
    band.perform()

    print("\n-- Band Stats --")
    print(f"Total band experience: {band.total_experience()} years")

    print("\n-- Validation Check --")
    try:
        bad_musician = Musician("Rookie", -3)
    except ValueError as e:
        print(f"Caught error: {e}")

    print("\n" + "=" * 40)
```
